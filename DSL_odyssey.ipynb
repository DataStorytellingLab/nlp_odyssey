## loading our libraries and dataset

```python
# importing our spacy and pandas libraries
import spacy
import pandas as pd

# loading up the model in english
nlp = spacy.load("en_core_web_sm")
```

```python
with open("wilson.txt", "r") as f:
    wilson_text = f.read()
    f.close()
```

Using a "slice" to see the first 100 characters

```python
wilson_text[:100]
```

Some light cleaning
- removing the "/n" character
- using python function `replace()`

```python
# replace '/n' with a space
wilson_text = wilson_text.replace('\n', ' ')   
```

```python
print(wilson_text[:100])
```

## the `Doc` object
When we pass the texts into the `nlp()` function, we transforming it into a `Doc` type of object in spaCy. Read more about the spaCy pipeline here: https://spacy.io/usage/spacy-101#pipelines

```python
wilson = nlp(wilson_text)
```

```python
# see more about Doc objects in spaCy documentation
# https://spacy.io/api/doc

type(wilson)
```

## `Token` attributes

```python
# each token has a number of attributes

for i in wilson[:10]:
  print(i.text)
```

## comparing nouns with conditional statements

```python
# load up pope text
with open("pope.txt", "r") as f:
    pope_text = f.read()
    f.close()

# cleaning /n from text
pope_text = pope_text.replace('\n', ' ')

# creating doc object
pope = nlp(pope_text)
```

```python
# filtering:
# using conditional statement to print out nouns

for i in wilson[:100]:
  if i.pos_ == "NOUN":
    print(i)
```

```python
for i in pope[:100]:
  if i.pos_ == "NOUN":
    print(i)
```

```python
# using conditional statement to print out verbs

for i in wilson[:100]:
  if i.pos_ == "VERB":
    print(i)
```

```python
for i in pope[:100]:
  if i.pos_ == "VERB":
    print(i)
```

## pulling out entities 
- let's focus on Wilson's text for now.
- we are going to pull out different entities.
- then we are going to explore a particular entity in depth.

```python
# what are entities?
wilson.ents[:10]
```

```python
# pull out entities

for i in wilson.ents[:10]:
  print(i, i.label_)
```

```python
# how would I pull out entities for just characters?

for i in wilson.ents[:100]:
    if i.label_ == "PERSON":
        print(i)
```

```python
for i in wilson.ents[:100]:
    if i.label_ == "PERSON":
        print(i.text, i.label_)
        print(i.sent)
```

## advanced filtering: `ent` + `pos_`

```python
for ent in wilson.ents:
    for token in ent:
        if token.text == 'Helen':
            if token.head.pos_ == "VERB":
                print("Entity:", ent.text)
                print("Verb:", token.head.text)
```

```python
for ent in fagles.ents:
    for token in ent:
        if token.text == 'Helen':
            if token.head.pos_ == "VERB":
                print("Entity:", ent.text)
                print("Verb:", token.head.text)
```

```python
for ent in pope.ents:
    for token in ent:
        if token.text == 'Helen':
            if token.head.pos_ == "VERB":
                print("Entity:", ent.text)
                print("Verb:", token.head.text)
```
