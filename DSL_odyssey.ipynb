```python
# importing our spacy and pandas libraries
import spacy
import pandas as pd

# loading up the model in english
nlp = spacy.load("en_core_web_sm")
```

```python
with open("wilson.txt", "r") as f:
    wilson_text = f.read()
    f.close()
```

Some light cleaning
- removing the "/n" character
- using python function `replace()`

```python
# replace '/n' with a space
wilson_text = wilson_text.replace('\n', ' ')   
```

```python
wilson_text[:100]
```

## the `Doc` object
When we pass the texts into the `nlp()` function, we transforming it into a `Doc` type of object in spaCy. Read more about the spaCy pipeline here: https://spacy.io/usage/spacy-101#pipelines

```python
wilson = nlp(wilson_text)
```

## comparing nouns with conditional statements
- we will load up another version of the text, by Alexander Pope.
- do you notice any differences between Wilson's and Pope's usage of nouns and verbs?

```python
# load up pope text
with open("pope.txt", "r") as f:
    pope_text = f.read()
    f.close()

# cleaning /n from text
pope_text = pope_text.replace('\n', ' ')

# creating doc object
pope = nlp(pope_text)
```

```python
# filtering:
# using conditional statement to print out nouns

for i in wilson[:100]:
  if i.pos_ == "NOUN":
    print(i)
```

```python
for i in pope[:100]:
  if i.pos_ == "NOUN":
    print(i)
```

```python
# using conditional statement to print out verbs

for i in wilson[:100]:
  if i.pos_ == "VERB":
    print(i)
```

```python
for i in pope[:100]:
  if i.pos_ == "VERB":
    print(i)
```

## back to `Doc`: pulling out `ents`
- let's focus on Wilson's text for now.
- we are going to pull out different entities.
- then we are going to explore a particular entity in depth.

```python
# what are entities?
wilson.ents[:10]
```

```python
# pull out entities

for i in wilson.ents[:10]:
  print(i, i.label_)
```

```python
# how would I pull out entities for just characters?

for i in wilson.ents[:100]:
    if i.label_ == "PERSON":
        print(i)
```

```python
# the ent object also has a property "sent"

for i in wilson.ents[:100]:
    if i.label_ == "PERSON":
        print(i.text, i.label_)
        print(i.sent)
```

## analyzing characterization with `ent` + `pos_`
- we will create advanced filters to get more information about characters
- Steps:
  1. find a partner
  2. choose a character from the story
  3. revise the conditional to filters out that character by name
  4. examine the results between two different versions of the story

```python
# load up fagles text
with open("fagles.txt", "r") as f:
    fagles_text = f.read()
    f.close()

# cleaning /n from text
fagles_text = fagles_text.replace('\n', ' ')

# creating doc object
fagles = nlp(fagles_text)
```

```python
# printing adjectives associated with Odysseus

for ent in wilson.ents:
    for token in ent:
        if token.text == 'Odysseus':
            for child in token.children:
                if child.pos_ == "ADJ":
                    print(ent.text, "->", child.text)
```

```python
# printing adjectives associated with Odysseus

for ent in fagles.ents:
    for token in ent:
        if token.text == 'Odysseus':
            for child in token.children:
                if child.pos_ == "ADJ":
                    print(ent.text, "->", child.text)
```
