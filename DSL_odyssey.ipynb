The goal of this workshop is to process, annotate, and analyze text in order to enable

To do so, we will be working with the Python programming langauge, and in particular, a package called [spaCy](https://spacy.io/), which was designed to work with text-based data. The dataset we will use comes from from Emily Wilson's translation of *The Odyssey*, 2017, and from Alexander Pope's translation, from 1725.

## loading our libraries and dataset

```python
# importing our spacy and pandas libraries
import spacy
import pandas as pd
import requests



# loading up the model in english
nlp = spacy.load("en_core_web_sm")
```

```python
with open("wilson.txt", "r") as f:
    wilson_text = f.read()
    f.close()
```

```python
with open("pope.txt", "r") as f:
    pope_text = f.read()
    f.close()
```

Using a "slice" to see the first 100 characters

```python
wilson_text[:100]
```

```python
pope_text[:100]
```

Some light cleaning
- removing the "/n" character
- using python function `replace()`

```python
# replace '/n' with a space
wilson_text = wilson_text.replace('\n', ' ')   
pope_text = pope_text.replace('\n', ' ')
```

```python
print(wilson_text[:100])
print(pope_text[:100])
```

## exploring linguistic annotations
When we pass the texts into the ~nlp()~ function, we transforming it into a ~doc~ type of object in spaCy.

```python
w_doc = nlp(wilson_text)   
p_doc = nlp(pope_text)
```

```python
# see more about Doc objects in spaCy documentation
# https://spacy.io/api/doc

type(w_doc)
```

We will use list slicing to explore sections (slices) of our dataset

```python
# now when we slice, we do so by words!

w_doc[:20]
```

```python
len(w_doc)
```

## loops!
We use loops to iterate through our dataset and pull out individual words. 

```python
for i in w_doc[:10]:
  print(i)
```

## attributes

```python
# each token has a number of attributes

for i in w_doc[:10]:
  print(i.text)
```

```python
# some other useful attributes, including part of speech, and whether it is an alpha character.
# https://spacy.io/api/token/#attributes 

for i in w_doc[:20]:
  print(i.text, i.pos_, i.is_alpha)
```

## comparing nouns with conditional statements

```python
# filtering:
# using conditional statement to print out nouns

for i in w_doc[:100]:
  if i.pos_ == "NOUN":
    print(i)
```

```python
for i in p_doc[:100]:
  if i.pos_ == "NOUN":
    print(i)
```

```python
# using conditional statement to print out verbs

for i in w_doc[:100]:
  if i.pos_ == "VERB":
    print(i)
```

```python
for i in p_doc[:100]:
  if i.pos_ == "VERB":
    print(i)
```

## pulling out phrases
- other aspects of doc: `.noun_chunks`

```python
# noun chunks with spacy

for chunk in w_doc[:100].noun_chunks:
  print(chunk.sent)
```

```python
# noun chunks with spacy

for chunk in p_doc[:100].noun_chunks:
  print(chunk)
  print(chunk.sent)
```

## pulling out entities 
- let's focus on Wilson's text for now.
- we are going to pull out different entities.
- then we are going to explore a particular entity in depth.

```python
# what are entities?
w_doc.ents[:10]
```

```python
# pull out entities

for i in w_doc.ents[:10]:
  print(i, i.label_)
```

```python
# how would I pull out entities for just characters?

for i in w_doc.ents[:10]:
    if i.label_ == "PERSON":
        print(i)
```

```python
for i in w_doc.ents[:100]:
    if i.label_ == "PERSON":
        print(i.text, i.label_)
        print(i.sent)
```

A more complex expression combines loops and if statements to filter out by certain people.

```python
for i in w_doc.ents:
    if i.text == "Helen":
        print(i.sent)
```

```python
for i in p_doc.ents:
    if i.text == "Helen":
        print(i.sent)
```

```python
w_doc[:1000].ents
```

```python
for ent in w_doc.ents:
    for token in ent:
        if token.text == 'Helen':
            if token.head.pos_ == "VERB":
                print("Entity:", ent.text)
                print("Verb:", token.head.text)
```

```python
for ent in w_doc.ents:
    for token in ent:
        if token.text == 'Zeus':
            if token.head.pos_ == "VERB":
                print("Entity:", ent.text)
                print("Verb:", token.head.text)
```
